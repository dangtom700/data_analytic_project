```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("daily_food_nutrition_dataset.csv")
data.head(10)
```

```python
# Check the duration of the dataset
data['Date'] = pd.to_datetime(data['Date'])
start_date = data['Date'].min()
end_date = data['Date'].max()
duration = end_date - start_date
print(f"Dataset duration: {duration.days} days from {start_date.date()} to {end_date.date()}")
```

```python
print(f"Largest user ID: {data["User_ID"].unique().max()}")
print(f"Smallest user ID: {data["User_ID"].unique().min()}")
```
