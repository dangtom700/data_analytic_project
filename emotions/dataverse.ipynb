```python
import pandas as pd
import os

# Visualization
import umap
import matplotlib.pyplot as plt
from sklearn.manifold import TSNE
from sklearn.decomposition import PCA

# Clustering
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Preprocessing
from sklearn.preprocessing import StandardScaler
```

```python
folder = "dataverse_package"
```

```python
def visualize_data(file):
    # Load and preprocess data
    file_path = os.path.join(folder, "data", file)
    data_df = pd.read_csv(file_path)

    # Drop Time column, extract Status
    data_df = data_df.drop("Time", axis=1)
    labels = data_df.pop("Status")
    label_ids = pd.factorize(labels)[0]  # Convert string labels to integers

    # Standardize features
    scaler = StandardScaler()
    data_scaled = scaler.fit_transform(data_df)

    # Dimensionality reduction
    umap_2d = umap.UMAP(n_components=2, random_state=0).fit_transform(data_scaled)
    tsne_2d = TSNE(n_components=2, random_state=0).fit_transform(data_scaled)
    pca_2d = PCA(n_components=2, random_state=0).fit_transform(data_scaled)

    # Plot results
    plt.figure(figsize=(15, 5))
    for i, (data_2d, title) in enumerate(zip([umap_2d, tsne_2d, pca_2d], ['UMAP', 't-SNE', 'PCA']), 1):
        plt.subplot(1, 3, i)
        plt.scatter(data_2d[:, 0], data_2d[:, 1], c=label_ids, cmap='tab10', s=10)
        plt.title(f"{title} - {file}")
        plt.xlabel("Component 1")
        plt.ylabel("Component 2")

    plt.tight_layout()
    plt.show()
    plt.savefig(os.path.join(folder, "img", f"{file.split('.')[0]}.png"))
```

```python
# Process all files in the data folder
header_list = {}
data_list = os.listdir(os.path.join(folder, "data"))
for file in data_list:
    if file.endswith(".csv"):
        headers = pd.read_csv(os.path.join(folder, "data", file), nrows=1).columns
        header_list[file] = headers
```

```python
# Headers in common between all files
common_headers = set.intersection(*map(set, header_list.values()))
print(common_headers)
```
