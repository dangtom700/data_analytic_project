```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("intelligent_indoor_environment_dataset.csv")
data.head(10)
```

```python
data.describe()
```

```python
# Convert to relative time measurement from the t0, format YYYY-MM-DD HH:mm:ss.######
data["timestamp"] = pd.to_datetime(
    data["timestamp"],
    format="%Y-%m-%d %H:%M:%S"
)

t0 = data["timestamp"].iloc[0]
data["time_seconds"] = (data["timestamp"] - t0).dt.total_seconds().astype(int)
data = data.drop(columns=['timestamp'])
data.head(10)
```
